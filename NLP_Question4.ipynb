```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
corpus=["ai ai ai hindi research","ai research ethics you","ai ethics ethics"]
```

```python
vector=TfidfVectorizer()
```

```python
tf=vector.fit_transform(corpus)
```

```python
tfdd=pd.DataFrame(tf.toarray(),columns=vector.get_feature_names_out())
```
